```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.max_info_columns', 200)

df = pd.read_parquet('../../data/processed/crimes_features.parquet')
df.head()
```

```python
crime_category_dount_df = (
    df
    .groupby('crime_category', as_index = False)
    .agg(count = ('crime_category', 'count')
        ).copy()
)

print(crime_category_dount_df.head())
```

```python
labels = crime_category_dount_df['crime_category'].str.capitalize().tolist()
sizes = crime_category_dount_df['count'].tolist()
colors = ['#2ca02c', '#ff7f0e', '#1f77b4']

fig, ax = plt.subplots(figsize=(5, 5))
wedges, texts, autotexts = ax.pie(
    sizes,
    labels=None,
    colors=colors,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'width': 0.4, 'linewidth': 2, 'edgecolor': 'white'},
    pctdistance=0.75
)

legend_elements = [
    plt.Rectangle((0, 0), 1, 1, color=colors[i], label=labels[i])
    for i in range(len(labels))
]
ax.legend(handles=legend_elements, loc='center', frameon=False, fontsize=10)
ax.set_title('Crime Category Distribution', fontsize=13)

plt.tight_layout()
#plt.savefig('../../outputs/00_additional_charts/person_vs_property_donut.png', dpi=150, bbox_inches='tight')
plt.show()
```
