## 4. Known data issues

Two irregularities in the record counts affect any year-over-year
comparison in this project:

- **2015–2016**: an anomalous drop followed by a spike, out of line with
  adjacent years
- **2024**: a sharp drop in record counts compared to previous years

No cause is established here: the LAPD documentation does not explain these
variations, and this project does not attempt to attribute them. They are
reported so that trends involving these years are read with the appropriate
caution.
